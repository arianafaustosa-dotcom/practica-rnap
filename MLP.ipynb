```python
import numpy as np
```

```python
def sigmoid(x):
  return 1 / (1 + np.exp(-x))

def sigmoid_derivada(x):
  s = sigmoid(x)
  return s * (1 - s)
```

```python
class MLP:
    def __init__(self, n_capas, lr=0.5):
        self.lr = lr
        self.w = []
        self.b = []

        for i in range(len(n_capas) - 1):
            w = np.random.randn(n_capas[i], n_capas[i+1]) * np.sqrt(2.0 / n_capas[i])
            b = np.zeros((1, n_capas[i+1]))
            self.w.append(w)
            self.b.append(b)

    def forward(self, x):
        self.A = [x]
        self.Z = []
        entrada = x

        for i in range(len(self.w)):
            z = np.dot(entrada, self.w[i]) + self.b[i]
            a = sigmoid(z)
            self.Z.append(z)
            self.A.append(a)
            entrada = a

        return entrada

    def backward(self, y_real):
        n = y_real.shape[0]

        delta = (self.A[-1] - y_real) * (self.A[-1] * (1 - self.A[-1]))

        for i in range(len(self.w)-1, -1, -1):
          dw = np.dot(self.A[i].T, delta) / n
          db = np.sum(delta, axis=0, keepdims=True) / n

        if i > 0:
            delta = np.dot(delta, self.w[i].T) * (self.A[i] * (1 - self.A[i]))

        self.w[i] -= self.lr * dw
        self.b[i] -= self.lr * db

    def entrenamiento(self, x, y, epocas=5000):
        for ep in range(epocas):
            salida = self.forward(x)
            self.backward(y)

            if ep % 200 == 0:
                print("epoca", ep)
                print("error", np.mean((salida - y)**2))

```

```python
x = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([0, 1, 1, 0])
```

```python
np.random.seed(42)

red = MLP(n_capas=[2, 4, 1], lr=0.5)

red.entrenamiento(x, y, epocas=2000)
```

```python
predicciones = red.forward(x)

for i in range(len(x)):
    print("Entrada:", x[i],
          "Predicción:", np.round(predicciones[i]),
          "Real:", y[i])
```
